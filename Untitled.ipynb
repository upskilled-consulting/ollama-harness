```python
import json
import re
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

# Core Definitions
WORD_RE = re.compile(r"\b\w+\b")
LINE_RE = re.compile(r'^\[(\d{2}):(\d{2}):(\d{2})\s*→\s*(\d{2}):(\d{2}):(\d{2})\]\s*(.+)$')
BAD_ENDINGS = {"in", "of", "and", "to", "the", "a", "an", "for", "with", "on", "at", "by", "from"}

def ts_to_sec(h, m, s): return int(h) * 3600 + int(m) * 60 + int(s)

def clean_text(text: str) -> str:
    return text.replace("\u00a0", " ").replace("“", '"').replace("”", '"').replace("’", "'").strip()

def word_count(text: str) -> int: return len(WORD_RE.findall(text))

def ends_badly(text: str) -> bool:
    words = [w.lower() for w in WORD_RE.findall(text)]
    return bool(words) and words[-1] in BAD_ENDINGS

def transcript_to_rows(transcript_path, audio_path, speaker_id="me"):
    rows = []
    for raw in Path(transcript_path).read_text(encoding="utf-8").splitlines():
        line = raw.strip()
        m = LINE_RE.match(line)
        if not m: continue
        sh, sm, ss, eh, em, es, text = m.groups()
        start, end = ts_to_sec(sh, sm, ss), ts_to_sec(eh, em, es)
        text = clean_text(text)
        if not text: continue
        rows.append({
            "audio_filepath": str(audio_path), "text": text, "start": start,
            "end": end, "duration": end - start, "speaker_id": speaker_id,
            "source_transcript": str(transcript_path),
        })
    return rows

def build_jsonl_dataset(transcripts_dir, notes_dir, out_dir, speaker_id="me"):
    out_dir = Path(out_dir); out_dir.mkdir(parents=True, exist_ok=True)
    all_rows = []
    for md in sorted(Path(transcripts_dir).glob("*-transcript.md")):
        stem = md.name.replace("-transcript.md", "")
        audio_path = Path(notes_dir) / f"{stem}.wav"
        if audio_path.exists():
            all_rows.extend(transcript_to_rows(md, audio_path, speaker_id=speaker_id))
    
    jsonl_path = out_dir / "manifest.jsonl"
    with jsonl_path.open("w", encoding="utf-8") as f:
        for row in all_rows: f.write(json.dumps(row, ensure_ascii=False) + "\n")
    return jsonl_path, Path(out_dir)/"manifest.csv", all_rows

def load_manifest(jsonl_path):
    return pd.DataFrame([json.loads(l) for l in Path(jsonl_path).open("r", encoding="utf-8") if l.strip()])

def filter_training_df(df, min_words=4, min_dur=1.5, max_dur=10.0):
    df = df.copy()
    df["text"] = df["text"].astype(str).map(clean_text)
    df["word_count"] = df["text"].map(word_count)
    df["bad_end"] = df["text"].map(ends_badly)
    return df[(df["duration"].between(min_dur, max_dur)) & (df["word_count"] >= min_words) & (~df["bad_end"])].drop(columns=["word_count", "bad_end"]).reset_index(drop=True)

def preview_rows(df, n=10): return df[["audio_filepath", "text", "duration"]].head(n)
```

```python
import torchaudio
import torch
import numpy as np
# SpeechBrain 1.0 renamed speakerrecognition -> speaker
from speechbrain.inference.speaker import EncoderClassifier

classifier = EncoderClassifier.from_hparams(
    source="speechbrain/spkrec-xvect-voxceleb",
    run_opts={"device": "cuda"},
)

def create_speaker_embedding(audio_np: np.ndarray) -> np.ndarray:
    with torch.no_grad():
        t = torch.tensor(audio_np).unsqueeze(0)  # [1, time]
        emb = classifier.encode_batch(t)
        emb = torch.nn.functional.normalize(emb, dim=2)
    return emb.squeeze().cpu().numpy()

# Smoke test
test_row = train_df.iloc[0]
wav, sr = torchaudio.load(test_row["audio_filepath"])
wav = wav[:, int(test_row["start"] * sr) : int(test_row["end"] * sr)]
if sr != 16000:
    wav = torchaudio.functional.resample(wav, sr, 16000)
emb = create_speaker_embedding(wav.squeeze().numpy())
print(f"Embedding shape: {emb.shape}")  # expect (512,)
```

```python
from dataclasses import dataclass
from typing import Any

TARGET_SR = 16000

def prepare_dataset(example):
    wav, sr = torchaudio.load(example["audio_filepath"])
    s = int(example["start"] * sr)
    e = int(example["end"]   * sr)
    wav = wav[:, s:e]
    if sr != TARGET_SR:
        wav = torchaudio.functional.resample(wav, sr, TARGET_SR)
    audio_np = wav.squeeze().numpy()

    text_in  = processor(text=example["text"], return_tensors="pt")
    audio_in = processor(audio_target=audio_np, sampling_rate=TARGET_SR, return_tensors="pt")

    example["input_ids"]          = text_in["input_ids"].squeeze()
    example["labels"]             = audio_in["labels"].squeeze()
    example["speaker_embeddings"] = create_speaker_embedding(audio_np)
    return example

drop_cols = ["audio_filepath", "start", "end", "duration",
             "speaker_id", "source_transcript", "text"]
processed = dataset.map(prepare_dataset, remove_columns=drop_cols)
print(processed)

@dataclass
class TTSDataCollatorWithPadding:
    processor: Any

    def __call__(self, features: list[dict]) -> dict:
        input_ids      = [{"input_ids": f["input_ids"]}      for f in features]
        label_features = [{"input_values": f["labels"]}      for f in features]

        batch        = self.processor.pad(input_ids, return_tensors="pt")
        labels_batch = self.processor.pad(audio_target=label_features, return_tensors="pt")

        labels = labels_batch["input_values"].masked_fill(
            labels_batch.attention_mask.unsqueeze(-1).ne(1), -100
        )
        batch["labels"]             = labels
        batch["speaker_embeddings"] = torch.tensor(
            np.array([f["speaker_embeddings"] for f in features]), dtype=torch.float32
        )
        return batch

from transformers import Seq2SeqTrainer, Seq2SeqTrainingArguments

training_args = Seq2SeqTrainingArguments(
    output_dir="./speecht5_nicho",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=1e-5,
    max_steps=500,
    fp16=True,
    logging_steps=10,
    save_strategy="steps",
    save_steps=100,
    predict_with_generate=True,
)

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=processed["train"],
    eval_dataset=processed["test"],
    data_collator=TTSDataCollatorWithPadding(processor=processor),
)

# trainer.train()
```

```python
from pathlib import Path
import json
import pandas as pd
from sklearn.model_selection import train_test_split

def load_manifest(jsonl_path):
    rows = []
    with Path(jsonl_path).open("r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                rows.append(json.loads(line))
    return pd.DataFrame(rows)

def filter_manifest(df, min_dur=1.5, max_dur=12.0):
    df = df.copy()
    df["text"] = df["text"].astype(str).str.replace(r"\s+", " ", regex=True).str.strip()
    df = df[df["duration"].between(min_dur, max_dur)]
    df = df[df["text"].str.len() >= 2]
    return df.reset_index(drop=True)

def split_train_valid(df, valid_frac=0.1, seed=42):
    if len(df) < 2:
        return df.copy(), df.iloc[0:0].copy()
    train_df, valid_df = train_test_split(df, test_size=valid_frac, random_state=seed)
    return train_df.reset_index(drop=True), valid_df.reset_index(drop=True)

def write_jsonl(df, path):
    path = Path(path)
    with path.open("w", encoding="utf-8") as f:
        for _, row in df.iterrows():
            f.write(json.dumps(row.to_dict(), ensure_ascii=False) + "\n")
```

```python
import re
from pathlib import Path
import pandas as pd

BAD_ENDINGS = {
    "in", "of", "and", "to", "the", "a", "an", "for", "with", "on", "at", "by", "from"
}

def word_count(text: str) -> int:
    return len(re.findall(r"\b\w+\b", text))

def ends_badly(text: str) -> bool:
    words = re.findall(r"\b\w+\b", text.lower())
    return bool(words) and words[-1] in BAD_ENDINGS

def filter_training_df(df: pd.DataFrame, min_words: int = 4, min_dur: float = 1.5, max_dur: float = 12.0):
    df = df.copy()
    df["text"] = df["text"].astype(str).str.replace(r"\s+", " ", regex=True).str.strip()
    df["word_count"] = df["text"].apply(word_count)
    df["bad_end"] = df["text"].apply(ends_badly)
    df = df[df["duration"].between(min_dur, max_dur)]
    df = df[df["word_count"] >= min_words]
    df = df[~df["bad_end"]]
    return df.drop(columns=["word_count", "bad_end"]).reset_index(drop=True)

def write_jsonl(df: pd.DataFrame, path):
    path = Path(path)
    with path.open("w", encoding="utf-8") as f:
        for _, row in df.iterrows():
            f.write(row.to_json(force_ascii=False) + "\n")
```
